# Amazon BestSelling Books Dataset 📚

### 1. - Instalar e Exportar o Pandas & Carregar o DataFrame

```python
import pandas as pd

df =pd.read_csv('../data/bestsellers with categories.csv')
```

### 2. - Explorar o DataFrame

```python
# Printa as 5 primeiras linhas do DataFrame
display(df.head())
# Printa o número de linhas e colunas do DataFrame
display(df.shape)
# Printa os nomes das colunas do DataFrame
display(df.columns)
# Printa o resumo estatístico do DataFrame
display(df.describe())
```

### 3. - Limpar e Organizar o DataFrame

```python
# Apagar os dados com os títulos duplicados
df.drop_duplicates(subset=['Name'], inplace=True)
# Verificar se há valores nulos em alguma coluna
display(df.isnull().sum())
```

```python
# Verificar a quantidade de valores duplicados apagados
display(df.shape)
#Comparando com o DataFrame original o valor apagado foi 550-351 = 199 linhas duplicadas.
print(f"Quantidade de Linhas Duplicadas Apagadas: {550-351}")
```

```python
#Renomear as colunas para português
df.rename(columns={'Name': 'Título', 'Author': 'Autor', 'User Rating': 'Avaliação do Usuário','Reviews': 'Avaliações', 'Price': 'Preço', 'Year': 'Ano', 'Genre': 'Gênero'}, inplace=True)
df['Gênero'] = df['Gênero'].replace({'Fiction': 'Ficção', 'Non Fiction': 'Não Ficção'})
display(df.head())
```

```python
# Converter o tipo de dado
df['Preço'] = df['Preço'].astype(float)
df['Preço_convertido'] = df['Preço'].apply(lambda x: f"$ {x:,.2f}")
display(df.dtypes)
display(df.head())
```

### 4. Realizar Análise de Dados

```python
# Titulos com mais avaliações
melhores_reviews = df.sort_values(by='Avaliações', ascending=False)
top_10_reviews =  melhores_reviews.head(10)
display(top_10_reviews[['Título', 'Autor', 'Ano', 'Avaliações']])
```

```python
# Títulos com melhores avaliações dos usuários
rating_usuário = df.sort_values(by='Avaliação do Usuário', ascending=False)
top_10_rating_usuário = rating_usuário.head(10)
display(top_10_rating_usuário[['Título', 'Autor', 'Ano', 'Avaliação do Usuário']])
# Houve um empate na avaliação do usuário, vamos quantificar quantos livros tem o valor 4.9
contagem_4_9 = df[df['Avaliação do Usuário'] == 4.9].shape[0]
print(f"A quantidade de títulos com avaliação de 4.9 é: {contagem_4_9} livros 📖")
```

```python
# Títulos com maior avaliação do usuário e mais avaliações
reviews_livros = df.sort_values(by=['Avaliação do Usuário', 'Avaliações'], ascending=[False, False])
top_10_reviews_livros = reviews_livros.head(10)
display(top_10_reviews_livros[['Título', 'Autor', 'Ano', 'Avaliação do Usuário', 'Avaliações']].reset_index(drop=True))
```

```python
# Popularidade do Autor
contagem_autor = df['Autor'].value_counts()
top_10_autores = contagem_autor.head(10)
top_10_autores.index = [f"{top_10_autores.index[0]} 🥇", f"{top_10_autores.index[1]} 🥈", f"{top_10_autores.index[2]} 🥉",] + list(top_10_autores.index[3:])
print("Autores mais populares:\n")
for autor, count in top_10_autores.items():
    print(f"{autor}: {count} livros 📖")
```

```python
# Qual ano teve mais livros publicados
contagem_anos = df['Ano'].value_counts()
print("Ano com mais livros publicados:\n")
for year, count in contagem_anos.items():
    print(f"{year}: {count} livros 📚")
```

```python
# Média de preço total dos livros
media_preço = df['Preço'].mean()
print(f"A média geral de preços dos livros é: $ {media_preço:.2f} 💵")
```

```python
# Média de preços por genero dos livros
media_preço_ficção = df[df['Gênero'] == "Ficção"].groupby('Gênero')['Preço'].mean().loc["Ficção"]
media_preço_não_ficção = df[df['Gênero'] == "Não Ficção"].groupby('Gênero')['Preço'].mean().loc["Não Ficção"]
print(f"Média de preços para Ficção: ${media_preço_ficção.round(2)} 💵")
print(f"Média de preços para Não Ficção: ${media_preço_não_ficção.round(2)} 💵")
```

```python
# Média de Avaliação por Gênero
media_avaliação_ficção = df[df['Gênero'] == "Ficção"].groupby('Gênero')['Avaliação do Usuário'].mean().loc["Ficção"]
media_avaliação_não_ficção = df[df['Gênero'] == "Não Ficção"].groupby('Gênero')['Avaliação do Usuário'].mean().loc["Não Ficção"]
print(f"Média de avaliações para Ficção: {media_avaliação_ficção.round(2)} ⭐")
print(f"Média de avaliações para Não Ficção: {media_avaliação_não_ficção.round(2)} ⭐")
```
